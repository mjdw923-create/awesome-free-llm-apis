```python
import time, requests
from google.colab import userdata

KEY = userdata.get("GEMINI_KEY")
URL = "https://generativelanguage.googleapis.com/v1beta/openai/chat/completions"
MODELS = ["gemini-3.5-flash", "gemini-3.8-flash", "gemini-3.1-flash-lite"]
H = {"Authorization": "Bearer " + KEY}

def call(m, p):
    b = {"model": m, "messages": [{"role": "user", "content": p}]}
    return requests.post(URL, headers=H, json=b, timeout=60)

def ask(p):
    for m in MODELS:
        for i in range(3):
            r = call(m, p)
            if r.ok: return m, r.json()["choices"][0]["message"]["content"]
            if r.status_code != 503: break
            time.sleep(2 ** i)
    return None, "failed"

m, a = ask("اكتب 3 عبارات ترحيب قصيرة بالعربية")
print(m)
print(a)
```
